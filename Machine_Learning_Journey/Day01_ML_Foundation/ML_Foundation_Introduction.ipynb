## 1. What is Machine Learning?

Machine Learning enables computers to learn patterns from data and use them to make predictions or decisions, instead of relying on explicitly written rules.

```
Data + Expected Results
        ↓
   ML Algorithm
        ↓
    ML Model
        ↓
Prediction / Decision
```

**Example:** Predicting house prices from historical data (area, bedrooms, location, age) instead of manually writing pricing rules.

## 2. AI vs ML vs Data Science vs Deep Learning

```
Artificial Intelligence
└── Machine Learning
    └── Deep Learning
```

- **AI** — systems that perform tasks requiring human-like intelligence
- **ML** — subset of AI; learns patterns from data
- **DL** — subset of ML; uses multi-layer neural networks
- **Data Science** — interdisciplinary field combining statistics, programming, ML, and visualization to extract insights; overlaps with AI/ML but isn't strictly a subset

## 3. Traditional Programming vs Machine Learning

| Traditional Programming | Machine Learning |
|---|---|
| Rules are explicitly written | Patterns are learned from data |
| Programmer defines logic | Algorithm learns relationships |
| Data + Rules → Output | Data + Output → Model |
| Good for well-defined rules | Good for complex patterns |
| Changes require modifying rules | Model can be retrained with new data |

## 4. Types of Machine Learning

```
Machine Learning
├── Supervised Learning      → learns from labeled data → Prediction
├── Unsupervised Learning    → learns from unlabeled data → Pattern discovery
└── Reinforcement Learning   → learns through interaction → Maximize reward
```

### 7. Supervised Learning
Learns from labeled data (Input Features + Known Target).

- **Regression** — predicts a continuous value (e.g., house price, salary)
- **Classification** — predicts a category (e.g., spam / not spam)

### 8. Unsupervised Learning
Works with unlabeled data to discover hidden patterns or groupings.

- **Clustering** — K-Means, Hierarchical, DBSCAN
- **Dimensionality Reduction** — PCA

### 9. Reinforcement Learning
An agent learns by interacting with an environment, receiving rewards or penalties, to maximize cumulative reward over time.

**Applications:** Robotics, game playing, autonomous systems, resource management.

## 10. Real-World Applications of Machine Learning

- **E-commerce** — recommendation systems
- **Banking & Finance** — fraud detection, credit scoring
- **Healthcare** — disease prediction, medical image analysis
- **Transportation** — route optimization, autonomous driving
- **Email** — spam classification

## 11. Practical Classification Exercise

Classifying five real-world problems by ML type.

````python
import pandas as pd

applications = {
    "Application": [
        "Email Spam Detection",
        "Customer Segmentation",
        "House Price Prediction",
        "Game Playing Agent",
        "Product Recommendation"
    ],
    "ML_Type": [
        "Supervised",
        "Unsupervised",
        "Supervised",
        "Reinforcement",
        "Unsupervised / Hybrid"
    ],
    "Problem_Type": [
        "Classification",
        "Clustering",
        "Regression",
        "Decision Making",
        "Recommendation"
    ]
}

df = pd.DataFrame(applications)
df
````

````python
df.info()
````

````python
df["ML_Type"].value_counts()
````

````python
df["Problem_Type"].value_counts()
````

````python
import matplotlib.pyplot as plt

df["ML_Type"].value_counts().plot(kind="bar")
plt.title("Real-World Applications by Machine Learning Type")
plt.xlabel("Machine Learning Type")
plt.ylabel("Number of Applications")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
````
