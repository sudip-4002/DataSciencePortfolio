Part:1 Univariate Analysis 

Univariate Analysis - Numeric feature

```python
sns.histplot(df['age'], kde = True)
plt.title("Distribution of age")
plt.show()
```

```python
sns.boxplot(x=df['age'])
plt.title('Boxplot of Age')
plt.show()
```

```python
sns.histplot(df['player_rating'], kde=True)
plt.title('Distribution of Player Rating')
plt.show()
```

```python
sns.boxplot(x=df['player_rating'])
plt.title('Boxplot of Player Rating')
plt.show()
```

Univariate Analysis - Categorical Feature

```python
sns.countplot(x='position', data=df, order=df['position'].value_counts().index)
plt.title('Position Count')
plt.show()
```

```python
sns.countplot(x='preferred_foot', data=df)
plt.title('Preferred Foot Count')
plt.show()
```

```python
sns.countplot(x='match_result', data=df)
plt.title('Match Result Distribution (W = Win, L = Loss, D = Draw)')
plt.show()
```

Bivariate Analysis


```python
sns.boxplot(x='position', y='player_rating', data=df)
plt.title('Player Rating by Position')
plt.show()
```

```python
sns.boxplot(x='preferred_foot', y='pass_accuracy', data=df)
plt.title('Pass Accuracy by Preferred Foot')
plt.show()
```

```python
sns.scatterplot(x='distance_covered_km', y='stamina_score', data=df)
plt.title('Distance Covered vs Stamina Score')
plt.show()
```

```python
sns.scatterplot(x='age', y='market_value_eur', data=df)
plt.title('Age vs Market Value')
plt.show()
```

Multivariate Analysis

```python
key_metrics = ['goals', 'assists', 'shots', 'expected_goals_xg', 'pass_accuracy',
               'tackles', 'interceptions', 'distance_covered_km', 'top_speed_kmh',
               'player_rating', 'performance_score', 'market_value_eur']

corr = df[key_metrics].corr(numeric_only=True)
corr
```

```python
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
sns.catplot(
    x='position',
    hue='preferred_foot',
    col='match_result',
    data=df,
    kind='count',
)
plt.show()
```

```python
sns.pairplot(
    df,
    hue='match_result',
    vars=['player_rating', 'pass_accuracy', 'distance_covered_km']
)
plt.show()
```

Final Conclusion

Forwards have the highest average player rating (3.88), followed closely by midfielders (3.86) and defenders (3.79); goalkeepers rate lowest (2.07) since the rating scale rewards attacking output.
Preferred foot (left vs right) makes almost no difference to pass accuracy (~80.7% vs ~80.9%).
Players rate slightly higher on average in matches their team won (3.69) compared to draws (3.60) or losses (3.61).
Age and market value show a very weak negative correlation, meaning younger and older players both appear across the market value range - age alone doesn't strongly predict a player's value.
Forwards score far more goals per match on average (0.14) than midfielders (0.05), defenders (0.02), and goalkeepers (~0).
Distance covered and stamina score move together, as expected physically.
