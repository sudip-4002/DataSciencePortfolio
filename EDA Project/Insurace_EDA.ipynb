Feature Extraction 

```python
from scipy.stats import pearsonr
import pandas as pd

selected_features = [
    'age',
    'bmi',
    'children',
    'is_female',
    'is_smoker',
    'region_northwest',
    'region_southeast',
    'region_southwest',
    'bmi_category_Normal weight',
    'bmi_category_Overweight',
    'bmi_category_Obesity'
]

correlation = {
    feature: pearsonr(df_cleaned[feature], df_cleaned['charges'])[0]
    for feature in selected_features
}

correlation_df = pd.DataFrame(
    list(correlation.items()),
    columns=['Feature', 'Pearson Correlation']
)

correlation_df.sort_values(
    by='Pearson Correlation',
    ascending=False,
    inplace=True
)

print(correlation_df)
```

```python
cat_features = [
    'is_female',
    'is_smoker',
    'region_northwest',
    'region_southeast',
    'region_southwest',
    'bmi_category_Normal weight',
    'bmi_category_Overweight',
    'bmi_category_Obesity'
]
```

```python
from scipy.stats import chi2_contingency
import pandas as pd

alpha = 0.05

# Create bins for target variable
df_cleaned['charges_bin'] = pd.qcut(df_cleaned['charges'], q=4, labels=False)

chi2_results = {}

# Perform Chi-Square Test
for col in cat_features:

    contingency = pd.crosstab(df_cleaned[col], df_cleaned['charges_bin'])

    chi2_stat, p_value, dof, expected = chi2_contingency(contingency)

    decision = (
        "Reject Null (Keep Feature)"
        if p_value < alpha
        else "Accept Null (Drop Feature)"
    )

    chi2_results[col] = {
        "chi2_statistic": chi2_stat,
        "p_value": p_value,
        "decision": decision
    }

# Create DataFrame AFTER the loop
chi2_df = pd.DataFrame(chi2_results).T

# Sort by p-value
chi2_df = chi2_df.sort_values(by="p_value", ascending=True)

# Display ONLY ONCE
display(chi2_df)
```
