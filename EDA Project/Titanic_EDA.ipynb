Visualization


```python
plt.figure(figsize=(10,5))
sns.heatmap(df.isnull(),cbar=False,cmap="viridis")
plt.title("Missing Values")
plt.show()
```

```python
df.duplicated().sum()
```

```python
df.nunique()
```

```python
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(8,6))
sns.heatmap(numeric_df.corr(),
            annot=True,
            cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

Part 1 : Univariate Analysis

Age Distribution

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Age"],
             bins=30,
             kde=True)

plt.title("Age Distribution")
plt.show()
```

Fare Distribution 

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Fare"],
             kde=True)

plt.show()
```

Passenger Class

```python
sns.countplot(data=df,x="Pclass")
plt.show()
```

Gender Distribution 

```python
sns.countplot(data=df,x="Sex")
plt.show()
```

Survival Count

```python
sns.countplot(data=df,x="Survived")
plt.show()
```

Embarked 

```python
sns.countplot(data=df,x="Embarked")
plt.show()
```

Outlier Detection 

```python
sns.boxplot(x=df['Age'])
plt.title('Boxplot of Age')
plt.show()
```

Bivariate Analysis

```python
sns.boxplot(x='Survived', y='Age', data=df)
plt.title('Age vs Survival')
plt.show()
```

```python
sns.boxplot(x='Survived', y='Fare', data=df)
plt.title('Fare vs Survival')
plt.show()
```

```python
sns.countplot(x='Sex', hue='Survived', data=df)
plt.title('Gender vs Survival')
plt.show()
```

```python
print(df.columns.tolist())
```

```python
sns.countplot(x='Pclass', hue='Survived',data=df)
plt.title('Class vs Survival')
plt.show()
```

```python
sns.scatterplot(x='Age',y='Fare',data=df)
plt.title('Age vs Fare')
plt.show()
```

Multivariate Analysis

```python
corr = df.corr(numeric_only=True)
corr
```

```python
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
sns.catplot(
    x='Pclass',
    hue='Sex',
    col='Survived',
    data=df,
    kind= 'count',
)
plt.show()
```

```python
sns.pairplot(
    df,
    hue='Survived',
    vars=['Age', 'Fare']
)
plt.show()
```

Final Conclusion
- Females survived more than males.
- Higher-class passengers had better chances of survival.
- Higher fare --> higher survival probability
- Children are more likely to survive.
- Third-class males passengers were the most vulnerable group.
