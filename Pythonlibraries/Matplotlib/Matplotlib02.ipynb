```python
data = [12, 15, 14, 10, 18, 22, 25, 17, 13, 19, 21, 16]

plt.hist(data, bins=5, color='steelblue', edgecolor='white')  # Plot frequency of values in 5 bins
plt.title('Histogram')   # Set chart title
plt.xlabel('Value')      # Label x-axis
plt.ylabel('Frequency')  # Label y-axis
plt.show() 
```

```python
data = [12, 15, 14, 10, 18, 22, 25, 17, 13, 19, 21, 16,26,28,11,31,34]

plt.hist(data, bins=[10,15,20,25,30,35], color='steelblue', edgecolor='white')  # Plot frequency of values in 5 bins
plt.xticks(np.arange(10,36,5))
plt.yticks( np.arange(1,6,1))
plt.title('Weight Distribution')   # Set chart title
plt.xlabel('Weight')      # Label x-axis
plt.ylabel('Frequency')  # Label y-axis
plt.show() 
```

Adding mean and median to histogram 

```python
data = [
    112, 145, 234, 178, 356, 423, 289, 467, 534, 612,
    723, 845, 956, 321, 456, 567, 678, 789, 890, 123,
    234, 345, 456, 567, 678, 789, 890, 901, 123, 234,
    315, 426, 537, 648, 759, 860, 971, 182, 293, 304,
    415, 526, 637, 748, 859, 960, 171, 282, 393, 404
]

plt.hist(data, bins=10 ,edgecolor="purple",color="green")

m = np.mean(data)
print(m)

plt.axvline(m,color="yellow")

n = np.median(data)
print(n)

plt.axvline(n , color="red", linestyle='--')
plt.show()
```

Advanced Histogram

Advanced Histogram adds statistical layers like KDE curve, mean/median lines, and density normalization to reveal deeper patterns beyond simple frequency counts.



```python
blood_pressure = [72, 65, 78, 81, 68, 74, 85, 61, 77, 83,
                  69, 76, 88, 63, 79, 84, 67, 73, 91, 58,
                  80, 66, 87, 71, 75, 82, 64, 89, 70]        # 29 values

blood_sugar    = [65, 78, 55, 82, 71, 60, 88, 74, 67, 91,
                  58, 76, 84, 62, 79, 53, 86, 69, 73, 95,
                  57, 81, 64, 77, 90, 54, 83, 68, 75]        # 29 values

cholesterol    = [75, 62, 85, 58, 79, 92, 66, 88, 71, 55,
                  83, 69, 95, 61, 77, 87, 53, 80, 64, 91,
                  57, 74, 86, 60, 78, 93, 67, 82, 70] 


plt.hist([blood_pressure, blood_sugar,cholesterol],bins =[50,60,70,80,90,100],color= ["orange", "red","green"],alpha= 0.6, edgecolor='blue')
plt.legend(["Pressure" , "Sugar" , "cholesterol"],loc = "upper left")
plt.xlabel("Sugar level")
plt.ylabel("Number of patients")
plt.xticks([50,60,70,80,90,100])
plt.yticks(np.arange(1,11,2))
plt.title("Blood Sugar Chart")
plt.show()        
```

```python
import matplotlib.pyplot as plt
import numpy as np

blood_pressure = [72, 65, 78, 81, 68, 74, 85, 61, 77, 83,
                  69, 76, 88, 63, 79, 84, 67, 73, 91, 58,
                  80, 66, 87, 71, 75, 82, 64, 89, 70]

blood_sugar    = [65, 78, 55, 82, 71, 60, 88, 74, 67, 91,
                  58, 76, 84, 62, 79, 53, 86, 69, 73, 95,
                  57, 81, 64, 77, 90, 54, 83, 68, 75]

cholesterol    = [75, 62, 85, 58, 79, 92, 66, 88, 71, 55,
                  83, 69, 95, 61, 77, 87, 53, 80, 64, 91,
                  57, 74, 86, 60, 78, 93, 67, 82, 70]

plt.hist([blood_pressure, blood_sugar, cholesterol],
          bins=[50,60,70,80,90,100],
          color=["orange", "red", "green"],
          alpha=0.6,
          edgecolor='white',
          orientation='horizontal')   # ✅ horizontal

plt.legend(["Pressure", "Sugar", "Cholesterol"], loc="upper right")
plt.xlabel("Number of patients")     # ✅ swapped
plt.ylabel("Sugar level")            # ✅ swapped
plt.xticks(np.arange(1,11,2))        # ✅ swapped
plt.yticks([50,60,70,80,90,100])     # ✅ swapped
plt.title("Blood Sugar Chart — Horizontal")
plt.show()
```

Overlaying bar plot on top of each other 

```python
# create data 
x =["A","B","C","D"]
y1 = np.array([10,20,10,30])
y2 = np.array([20,25,15,25])
y3 = np.array([12,15,19,6])
y4 = np.array([10,29,13,19])

# plot bars in stack manner 
plt.bar(x, y1 , color="red")
plt.bar(x, y2 ,bottom=y1 ,color="blue")
plt.bar(x, y3 , bottom = y1+y2, color="green")
plt.bar(x, y4 , bottom = y1+y2+y3, color="yellow")

plt.xlabel("Teams")
plt.ylabel("Score")
plt.legend(["Round1","Round2","Round3","Round4"])
plt.title("Score by Teams upto 4 Rounds")
plt.show()
```

```python
# Horizontal bar graph 
# create data 
x =["A","B","C","D"]
y1 = np.array([10,20,10,30])
y2 = np.array([20,25,15,25])
y3 = np.array([12,15,19,6])
y4 = np.array([10,29,13,19])

# plot bars in stack manner 
plt.barh(x, y1 , color="red")
plt.barh(x, y2 ,left=y1 ,color="blue")
plt.barh(x, y3 , left = y1+y2, color="green")
plt.barh(x, y4 , left = y1+y2+y3, color="yellow")

plt.xlabel("Score")
plt.ylabel("Teams")
plt.legend(["Round1","Round2","Round3","Round4"])
plt.title("Score by Teams upto 4 Rounds")
plt.show()
```

Box plot

Box Plot displays the minimum, Q1, median, Q3, and maximum of data in a single box shape to show spread, center, and outliers at a glance.



```python
x = [10,20,30,40,50,60,70]
plt.boxplot(x)
plt.show()
```
