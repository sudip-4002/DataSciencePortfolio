np.arange()function in numpy works like python range()to create sequences of numbers. its syntax : np.arange (start, end step)

Start(optional):The stating number(default is 0).

end: The ending number (not included in the array).

step(optional): The diffence between number(default is 1).

```python
array1 = range(10)
print(list(array1))
array2 = range(3,20,5)
print(list(array2))
array3 = range(3,20,2.5) # in range function we didn't use float value
print (array3)
```

```python
# create array with valus from 0 to 9
array = np.arange(10)
print(array)

# create an array with valus form 3 to 20 with step of 2.3
array1 = np.arange(3,20,2.3)
print(array)
```

np.linespace()function creates an array with evenly spaced numbers within a range. it's syntax is : np.linespace(start,end,num)

start:Starting value of the range.

end : Ending value of the range(inclusive).

num(optional): Number of values to create

```python
print(np.linspace(0,10,5))
```

```python
print(np.linspace(0,500,100))
```

```python
# elements between 10 and 100 with stepsize 4
array1 = np.arange(10,100,4)

# generate 4 elements between 10 and 100
array2 =np.linspace(10,100,4)

print("Using arange:",array1)#doesn't include 100
print("Using linspace:",array2)#includes 100
```
